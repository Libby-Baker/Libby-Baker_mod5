## Task 1 — Sentinel stats (≈15 min)

Write a program that repeatedly asks the user for **integers between 0 and 100**.  
Use a **sentinel**: the user types `done` (any case) to finish.

**Requirements**
- Ignore blank inputs (reprompt).
- If the user types something that's not an integer in range, print a friendly message and reprompt.
- When the user finishes, print: **count**, **sum**, **min**, **max**, and **mean** (to 2 decimals) using f-string: `{num:.2f}`.
- If no valid numbers were entered, print `"No data."`

````python
numbers = [] #set up list to be filled by the function below
while True:# Loop needs to be broken deliberately, in this case after the word done
  number = input("Enter a number from 0 to 100 or done to stop ").strip().lower()#Gets and normalizes input for numbers
  if number == "done":#checks if input was "done"
    break#If input was done, end loop
  else:#If input was not done
    try:#Try converting it into an integer
      number = int(number)
    except ValueError: #If input could not be converted
      print("Input must be a number") #ValueError
      continue #Skip input and go again
    else: #If input was successfully converted
      if number < 0 or number > 100: #make sure it's within the range or 0-100
       print("Number must be between 0 and 100 ") #If not, print message
       continue #discard input, go again
      else: #If number was within range
        numbers.append(number) #add it to the list of numbers
#after the loop
if len(numbers) == 0:#If no valid data was entered
  print ("No data")#No data
else: #Otherwise
  print(f"You entered {len(numbers)} numbers.") #Print total amount of numbers entered
  print(f"Sum: {sum(numbers)}") #print sum of numbers
  print(f"Lowest number: {min(numbers)}") #print lowest number
  print(f"Highest number: {max(numbers)}") #print highest number
  print(f"Mean: {sum(numbers)/len(numbers):.2f} ") #print average
````

## Task 2 — Choice validator (≈8 min)

Write a function `get_choice(prompt, options)` that:
- Prints the prompt and the list of options.
- Repeatedly asks for input until the user types **exactly** one of the options (case-insensitive).
- Returns the **originally cased** option from the `options` list.

**Example**
```text
Your choice (red/blue/green): Blue
# returns "blue" if options = ["red", "blue", "green"]
```

> Hint: Build `options_lower` once, then search it each time to find the index.

````python
def get_choice(prompt, options): #define function for color input, with the prompt and options as arguements
  print(prompt) #print the prompt
  color = input(options) #ask for input with options
  return(color)

color = get_choice("Pick a color", "red/blue/green ") #call the color input function, with prompt and options
#Loop to validate the inputs: Until the color input, in lowercase, matches one of the choices, the user will be prompted to answer again
while not(color.lower() == "red" or color.lower() == "green" or color.lower() == "blue"):
  color = get_choice("Pick a color", "red/blue/green ")
print("You chose:", color)#Once the input is acceptable, print it
````

## Task 3 — Multiplication grid (≈12 min)

Prompt the user for two **integers** `rows` and `cols` in the range **1..10**.  
Validate both with try/except and reprompt until valid.

Then print a **rows × cols** multiplication table, where each cell shows `r*c` (1-indexed).  
Format with spaces so columns line up **at least** for small sizes.

**Example (rows=3, cols=4)**
```
    1  2  3  4
 1: 1  2  3  4
 2: 2  4  6  8
 3: 3  6  9 12
```

````python
rows = 0#create variabel rows to be manipulated by loop
col = 0 #create variable collumns
while not((rows > 1 and rows < 10) and (col > 1 and col < 10)): #While the values of both the rows and collumns are not in the acceptable range, keep prompting for them
  rows_ans = input("Enter a number from 1 - 10 for rows ") #get input for rows
  col_ans = input("Eneter a number from 1 - 10 for collumns ") #get input for collumns
  try: #try casting inputs to integers
    rows = int(rows_ans)
    col = int(col_ans)
    if (rows < 1 or rows > 10) or (col < 1 or col > 10):#If the numbers are out of range
      print("Please enter a number between 1 and 10 for both rows and collumns") #say so
      continue #discard inputs ad reprompt
  except ValueError:#If the inputs can't be converted to integers at all
    print("Please enter an integer for both inputs") #ValueError
    continue #discard inputs and reprompt
  else: #Ony if both inputs are acceptable
    for r in range(1,rows + 1): #for each row of the table
      for c in range(1,col + 1): #for every collumn
       print(f"{c*r}", end=" ") #print the collumn times the row number, with a space between each collumn
      print() #print a space at the ned of each row


````

## Task 4 — Vowel counter across words (≈15 min)

Use a **sentinel** loop to read words until the user types `done`.  
For each word, use a **nested loop** (loop over characters) to count **vowels** `a e i o u` (case-insensitive).

**Output**
- After each word, print `"{word}" has {k} vowel(s)`.
- At the end, print the **total number of vowels** across all words.

> Hint: A string is looped over the way a list is.  Use `for ch in word:` and compare `ch.lower()` against a set like `{'a','e','i','o','u'}`.

````python
def get_word():#function to collect word inputs
  word = input("Enter a word ")
  return(word)

vowels = ["a", "e", "i", "o", "u"] #creat a list of vowels for the loop to compare to
vowel_count = 0 #create vowel count variable for loop to add to

word = get_word() #get the first word

while word.lower() != "done": #the sentinel word is done, so while the input is not the word done, run the loop
  vowel_per_word = 0 #create variable to track vowels of indevidual words
  for ch in str(word): #for each character in the word
    for vowel in vowels: #run through the list of vowels
      if ch.lower() == vowel: #If the character, in lowercase, matches a vowle on the list
        vowel_count += 1 #add a vowel to the total vowel count
        vowel_per_word += 1 #And add a vowel to the word vowel count
      else: #If the character is not a vowel
        continue #move on
  print(f"{word} has {vowel_per_word} vowels") #Once we've gone through the word,print how many vowels it has
  word = get_word() #get the next word

print(f"You entered {vowel_count} vowels") #Once we've exitted the loop, print the total amount of vowles entered


````

## Task 5 — Mini-gradebook (≈20 min)

Build a tiny gradebook using **both** a sentinel loop and a nested loop.

**Requirements**
1. Repeatedly ask for a **student name** until the user types `done` (sentinel).  
   - Ignore blank names.
2. For each student, read exactly **3 scores** (0..100, integers).  
   - Use input validation with try/except; reprompt on bad input.
3. Store each student's average in a dictionary `grades[name] = avg`.  **OR** use two parallel lists instead.
4. THIS REQUIREMENT IS DELETED +3 points if you have coded this : After input ends, print a simple report sorted by name, e.g.:
   ```
   Alice : 88.00
   Bob   : 92.33
   ```

Note: because dictionaries were not yet discussed, you are welcome to use **two different lists** instead.  
One list stores the names and the other stores the averages.  For example:
`list_1[0] == 'Alice'`, `list_1[1] == 'Bob'` and `list_2[0] == 88.00`, `list_2[1] == 92.33`.
To store data:
`list_1.append('Alice')` and `list_2.append(88.00)`.

**Stretch (optional)**  
- Also print the **class average**.

````python
students = [] #create list students to add to
student_averages = [] #create parallel list of student averages to add to
class_average = 0 #create variable class average to total up

def get_students(): #create function to get names, and validate them
  name = input("Enter a name ") #get name
  while name == "" or name == " ": #validation loop: if the name input is blank
    name = input("Enter a name ") #repromt for a new name
  return(name)


name = get_students()#get first name

while name.strip().lower() != "done":#sentinel loop: word done. While the name is not done, run loop
  attempts = [] #attempts list for average and while loop
  while len(attempts) < 3: #While the amount of attempts is less than 3
    score = input(f"Please enter score #{len(attempts) + 1}: ")#Prompt for a score input
    try:#try converting score input into int
      score = int(score)
      if score < 0 or score > 100:#If score is not in range
        print("Score must be a number from 0 to 100") #print message
        continue #discard input and repeompt
    except ValueError: #if score can't be converted
      print("Score must be a number") #print a message
      continue #discard input and reprompt
    else: #if input was valid
      attempts.append(score) #add it to the list of scores
  #At the end of each student
  students.append(name) #append the student's name to the list
  average = sum(attempts)/3 #calculate the student's average
  student_averages.append(average) #Add student's average
  class_average += average #Add student's average to the class average
  print(f"{name}: {average:.2f}") #Print student's average
  name = get_students() #get next student's name

class_average = (class_average/len(students)) #after exitting the whole loop, calculate the class average by dividng the totaled averages by the amount of students


print(students) #Print list of students
print(student_averages) #print list of student's averages
print(f"Class average: {class_average:.2f}") #print total class average







````
